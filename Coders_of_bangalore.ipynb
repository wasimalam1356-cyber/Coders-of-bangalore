```python
with open("final_data.txt", encoding = "utf=8") as f:
    data = f.read()
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)> 3]
```

```python
def parse_chunk(chunk):
    chunk = chunk.strip()
    lines = chunk.split("\n")
    username = lines[0]
    no_of_posts = int(lines[1].split(" post")[0].replace(",","").strip())
    no_of_followers = float(lines[2].split(" follower")[0].replace(",", "").replace("K","").replace("M", ""))
    if "K" in lines[2]:
        no_of_followers = int(no_of_followers * 1000)
    elif "M" in lines[2]:
        no_of_followers = int(no_of_followers * 1000000)
    else:
        no_of_followers = int(no_of_followers)
    no_of_following = int(lines[3].split(" followin")[0].replace(",", ""))
    name = lines[4]
    if(len(lines)>5):
        type_of_page = lines[5]
        bio = lines[6:]
    else:
        type_of_page = "Unknown"
        bio = ""

    return {"username" : username,
            "no_of_posts": no_of_posts,
            "no_of_followers":no_of_followers,
            "no_of_following":no_of_following,
            "name":name,
            "type_of_page":type_of_page,
            "bio": bio}
    
```

```python
parse_chunk(chunks[0])
```

```python
all_chunks= []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json

s = json.dumps(all_chunks, indent=4)
with open("data.json", "w") as f:
    f.write(s)
```

# Who has the maximum posts?

```python
max = 0
for chunk in all_chunks:
    if(max<chunk["no_of_posts"]):
        max= chunk["no_of_posts"]
        chunk_with_max_post = chunk
print(f"\"{chunk_with_max_post["username"]}\" has the highest posts.")
```

# Who has maximum followers

```python
highest =0
for chunk in all_chunks:
    if(highest<chunk["no_of_followers"]):
        highest= chunk["no_of_followers"]
        chunk_with_highest_followers= chunk
print(f"\"{chunk_with_highest_followers["username"]}\" has the highest followers")
```

# Who follows maximum people

```python
most =0
for chunk in all_chunks:
    if(most<chunk["no_of_following"]):
        most= chunk["no_of_following"]
        chunk_with_highest_following= chunk
print(f"\"{chunk_with_highest_following["username"]}\" has highest following")
```
